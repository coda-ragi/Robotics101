## Overview
This notebook contains a sample program demonstrating how to use **NumPy**, **SciPy**, and **Matplotlib** to model and simulate a robotic system. 

### Problem Statement:
Simulate a robot moving in a 2D environment toward a target goal and determine how close the robot reached the goal.

### Core Ecosystem Stack Used:
* **NumPy**: Handles robot data using numerical arrays and matrix operations.
* **SciPy**: Provides scientific and numerical algorithms (specifically calculating Euclidean distance).
* **Matplotlib**: Handles data visualization to plot the robot's trajectory.

## Slide 41: Initializing Packages and Robot Variables

In this phase, the necessary libraries are imported and the initial constraints of the simulation are defined:
* The initial starting pose of the robot ($X=0, Y=0$).
* The target coordinate goal ($X=5, Y=4$).
* An array of incremental movement vectors `(dx, dy)` representing the robot's steps.
* A Python list initialized with the starting pose to record the trajectory.

```python
# Import necessary packages
import numpy as np
from scipy.spatial.distance import euclidean
import matplotlib.pyplot as plt

# Robot starting position
pose = np.array([0, 0])

# Goal position
goal = np.array([5, 4])

# Robot movements (dx, dy)
motions = np.array([[1, 1], [1, 0.5], [1, 0.5], [1, 1]])

# Store robot path by initiating a list
path = [pose]

print("Initial Pose:", pose)
print("Goal Position:", goal)
print("Motion Steps:\n", motions)
```

## Slide 42: Execution, Analysis, and Visualization

In this step, the simulation loop executes:
1. **Update Position**: The loop iterates through each movement step, recursively adding the displacement vector to the robot's current pose.
2. **Log Path**: The updated pose is appended to the path list, which is subsequently converted into a NumPy matrix array for structural slicing.
3. **Compute Final Distance**: SciPy's `euclidean()` utility analyzes the spatial distance between the last element of the path matrix (`path[-1]`) and the goal.
4. **Plot Trajectory**: Matplotlib draws a visual layout of the map environment, mapping the discrete coordinate steps against the goal target point.

```python
# Update robot position by looping through the movement steps
for m in motions:
    pose = pose + m
    # Append the updated pose to the path
    path.append(pose)

# Convert list to numpy array
path = np.array(path)

# Compute final Euclidean distance to goal using SciPy
dist = euclidean(path[-1], goal)
print(f"Final Pose: {path[-1]}")
print(f"Euclidean distance to the goal: {dist:.4f} meters")

# Visualize robot path using matplotlib
plt.figure(figsize=(7, 5))
plt.plot(path[:, 0], path[:, 1], '-o', label="Robot Path", markersize=8)
plt.scatter(goal[0], goal[1], color='red', s=100, zorder=5, label="Goal")

# Graph Formatting
plt.title("Robot Movement")
plt.xlabel("X position")
plt.ylabel("Y position")
plt.xlim(-0.5, 6.0)
plt.ylim(-0.5, 5.0)
plt.grid(True)
plt.legend()
plt.show()
```
